```python
import pretty_midi

midi_data = pretty_midi.PrettyMIDI("mozart-eine-kleine-nachtmusik-allegro.mid")
```

```python
notes = []

for instrument in midi_data.instruments:
    if instrument.is_drum:
        continue
    
    for note in instrument.notes:
        notes.append((note.start, note.pitch))
```

```python
notes.sort(key=lambda x: x[0])

pitches = [n[1] for n in notes]

print(pitches[:20])
```

```python
from collections import defaultdict
import random

transitions = defaultdict(list)

for i in range(len(pitches) - 1):
    transitions[pitches[i]].append(pitches[i+1])

current = pitches[0]
generated = [current]

for _ in range(80):
    current = random.choice(transitions[current])
    generated.append(current)

print(generated)
```

```python
import pretty_midi

new_midi = pretty_midi.PrettyMIDI()
instrument = pretty_midi.Instrument(program=73)  
# 73 = flute（标准GM音色）

time = 0
duration = 0.25

for pitch in generated:
    note = pretty_midi.Note(
        velocity=90,
        pitch=pitch,
        start=time,
        end=time + duration
    )
    instrument.notes.append(note)
    time += duration

new_midi.instruments.append(instrument)
new_midi.write("generated_flute.mid")
```
