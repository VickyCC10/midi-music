```python
import pretty_midi

midi_data = pretty_midi.PrettyMIDI(
    "/Users/zhouhaiying/Downloads/f-f-chopin-prelude-op-28-no-4.mid"
)
```

```python
notes = []

for instrument in midi_data.instruments:
    for note in instrument.notes:
        notes.append(note.pitch)

print(notes[:20])
```

```python
from collections import defaultdict
import random

transitions = defaultdict(list)

for i in range(len(notes) - 1):
    transitions[notes[i]].append(notes[i+1])

current = notes[0]
generated = [current]

for _ in range(50):
    current = random.choice(transitions[current])
    generated.append(current)

print(generated)
```

```python
import pretty_midi

new_midi = pretty_midi.PrettyMIDI()
instrument = pretty_midi.Instrument(program=0)

time = 0
duration = 0.3

for pitch in generated:
    note = pretty_midi.Note(
        velocity=100,
        pitch=pitch,
        start=time,
        end=time + duration
    )
    instrument.notes.append(note)
    time += duration

new_midi.instruments.append(instrument)
new_midi.write("generated.mid")
```
